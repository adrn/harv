```python
import astropy.table as at
import matplotlib.pyplot as plt
import numpy as np
import pathlib
from astropy.io import ascii

%matplotlib inline
```

```python
allstar = at.Table.read("~/data/APOGEE/DR17/allStarLite-dr17-synspec.fits")
allvisit = at.Table.read("~/data/APOGEE/DR17/allVisitLite-dr17-synspec.fits")
```

## General search for low-epoch sources:

```python
stars = allstar[allstar["NVISITS"] == 6]

rng = np.random.default_rng(123)
idx = np.arange(len(stars))
rng.shuffle(idx)

stars = stars[idx]
stars = stars[(stars["STARFLAG"] == 0) & (stars["LOGG"] > 1.5) & (stars["LOGG"] < 3.5)]
stars = stars[(stars["VSCATTER"] > 0.5) & (stars["VSCATTER"] < 5.0)]
len(stars)
```

```python
star = stars[15]
visits = allvisit[allvisit["APOGEE_ID"] == star["APOGEE_ID"]]
visits = visits[visits["RV_FLAG"] == 0]

plt.errorbar(
    visits["JD"], visits["VHELIO"], yerr=visits["VRELERR"], marker="o", ls="none"
)
```

```python
visits["JD", "VHELIO", "VRELERR"].write(
    f"apogeedr17-{star['APOGEE_ID'].strip()}.csv", format="ascii.csv", overwrite=True
)
```

```python
xm_mask = np.isin(allstar["APOGEE_ID"], xm["APOGEE"])
harps_stars = at.unique(allstar[xm_mask], keys="APOGEE_ID")
harps_stars = harps_stars[harps_stars["NVISITS"] > 3]

harps_stars = at.join(
    harps_stars, xm, keys_left="APOGEE_ID", keys_right="APOGEE", join_type="inner"
)
harps_stars = at.unique(harps_stars, keys="APOGEE_ID")
harps_stars.sort("NVISITS", reverse=True)

len(harps_stars)
```
