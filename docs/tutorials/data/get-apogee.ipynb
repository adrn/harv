```python
import astropy.table as at
import matplotlib.pyplot as plt
import numpy as np

%matplotlib inline
```

```python
allstar = at.Table.read("~/data/APOGEE/DR17/allStarLite-dr17-synspec.fits")
allvisit = at.Table.read("~/data/APOGEE/DR17/allVisitLite-dr17-synspec.fits")
```

## General search for low-epoch sources:

```python
stars = allstar[allstar["NVISITS"] == 6]

rng = np.random.default_rng(123)
idx = np.arange(len(stars))
rng.shuffle(idx)

stars = stars[idx]
stars = stars[(stars["STARFLAG"] == 0) & (stars["LOGG"] > 1.5) & (stars["LOGG"] < 3.5)]
stars = stars[(stars["VSCATTER"] > 0.5) & (stars["VSCATTER"] < 5.0)]
len(stars)
```

```python
star = stars[15]
visits = allvisit[allvisit["APOGEE_ID"] == star["APOGEE_ID"]]
visits = visits[visits["RV_FLAG"] == 0]

plt.errorbar(
    visits["JD"], visits["VHELIO"], yerr=visits["VRELERR"], marker="o", ls="none"
)
```

```python
visits["JD", "VHELIO", "VRELERR"].write(
    f"apogeedr17-{star['APOGEE_ID'].strip()}.csv", format="ascii.csv"
)
```

```python
import os

from astropy.io import ascii
```

```python
kounkel_cat = ascii.read(
    os.path.expanduser("~/data/APOGEE/vac/Kounkel2021/table1.dat"),
    format="fixed_width",
    names=[
        "ID",
        "RAdeg",
        "DEdeg",
        "SBn",
        "Nepoch",
        "qW",
        "e_qW",
        "RVelW",
        "e_RVelW",
        "Per",
        "e_Per",
        "T0",
        "e_T0",
        "e",
        "e_e",
        "omega",
        "e_omega",
        "RVel",
        "e_RVel",
        "K1",
        "e_K1",
        "K2",
        "e_K2",
        "M1sin3i",
        "e_M1sin3i",
        "M2sin3i",
        "e_M2sin3i",
        "asini",
        "e_asini",
        "MaxdRV",
        "AmpRV1",
        "AmpRV2",
        "MaxT",
        "fLOS",
        "fVar",
        "PerTESS",
    ],
)
sub_tbl = kounkel_cat[
    (kounkel_cat["SBn"] == 2)
    & (kounkel_cat["Nepoch"] >= 6)
    # & (kounkel_cat["Nepoch"] < 10)
    & (kounkel_cat["e"] > 0.1)
    & ~kounkel_cat["K1"].mask
    & ~kounkel_cat["K2"].mask
    & (kounkel_cat["K2"] < 0.9 * kounkel_cat["K1"])
]
sub_tbl
```
